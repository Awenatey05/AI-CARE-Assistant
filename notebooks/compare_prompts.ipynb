### Check: all four prompts were given the same 50 reports, in the same order

```python
results = pd.read_csv("evaluation/prompt_experiment/results.csv")
cases_per_prompt = {prompt: list(group["case_id"]) for prompt, group in results.groupby("prompt", sort=False)}

for prompt, cases in cases_per_prompt.items():
    print(prompt, len(cases))

lists = list(cases_per_prompt.values())
print("Same reports, same order:", all(cases == lists[0] for cases in lists))
```
